```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
%load_ext autoreload
%autoreload 2
from streameucliddes.analysis.data_manipulation import get_stream_snr,scan_snr_grid
from streameucliddes.plotting.plotting import plot_snr_contours, plot_snr_maps
from streameucliddes.plotting import plot_style
```

```python
fig_folder = "../figures/outputs"
```

```python
sb_tests = np.linspace(33., 37, 20)
dm_tests = np.linspace(15., 20., 20)
axis_values = {"surface_brightness": sb_tests, "distance_modulus": dm_tests}
seed = 42


match_filter_list = [
    {
        "name": "g-r",
        "reference_band": ("des_yr6", "g"),
        "color_bands": [("des_yr6", "r")],
    },
        {
        "name": "VIS-H",
        "reference_band": ("euclid_q1", "VIS"),
        "color_bands": [("euclid_q1", "H")],
    },
            {
        "name": "g-H",
        "reference_band": ("des_yr6", "g"),
        "color_bands": [("euclid_q1", "H")],
        "red_cap_value": 2.0,
    },
                {
        "name": "r-h",
        "reference_band": ("des_yr6", "r"),
        "color_bands": [("euclid_q1", "H")],
    },
    {
        "name": "g-r-H",
        "reference_band": ("des_yr6", "g"),
        "color_bands": [("euclid_q1", "H"),("des_yr6", "r")],
    },
    {
        "name": "VIS-Y",
        "reference_band": ("euclid_q1", "VIS"),
        "color_bands": [("euclid_q1", "Y")],
    },
    {
        "name": "g-VIS",
        "reference_band": ("des_yr6", "g"),
        "color_bands": [("euclid_q1", "VIS")],
    },
]

bands = {"euclid_q1": ["VIS", "H","Y"],"des_yr6": ["g", "r"],}
```

```python
results_snr = scan_snr_grid(seed=seed, axis_values=axis_values,match_filter_list=match_filter_list,bands=bands)
```

```python
# SNR = 5 contour of each match filter: one color each, legend + label on each
# contour
#names_to_plot=None
names_to_plot = ["g-r", "VIS-H", "g-H"]
names_to_plot = ["g-r", "VIS-H", "g-H", "g-r-H","g-VIS"]
fig, ax = plt.subplots()
fig, ax = plot_snr_contours(results_snr, snr_level=5, names_to_plot=names_to_plot, ax=ax,)
```

```python
fig.savefig(os.path.join(fig_folder, f"snr_matchfilters.png"), dpi=300, bbox_inches="tight")
```
